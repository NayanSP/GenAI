```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_classic.document_loaders import TextLoader
from langchain_classic.text_splitter import CharacterTextSplitter
from langchain_openai import OpenAIEmbeddings
from langchain_classic.chains import RetrievalQA
from langchain_chroma import Chroma
from langchain_classic.embeddings.sentence_transformer import SentenceTransformerEmbeddings
```

```python
load = TextLoader('Sample.txt')
docs = load.load()
docs
```

```python
docs
```

```python
len(docs)
```

```python
text_split = CharacterTextSplitter(chunk_size=200, chunk_overlap = 0)
texts = text_split.split_documents(docs)
```

```python
len(texts)
```

```python
texts
```

```python
embed = SentenceTransformerEmbeddings(model_name="all-MiniLM-L6-v2")
```

```python
db = Chroma.from_documents(texts, embed)
```

```python
results = db.get(include=["documents", "embeddings"])
```

```python
results
```

```python
retriever = db.as_retriever(search_kwargs={"k":2})
retriever
```

```python
results = retriever.invoke("What is capital of India")
results
```

```python
for doc in results:
    print(doc.page_content)
    print(doc.metadata)
```
