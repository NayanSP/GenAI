```python
from langchain_openai import ChatOpenAI
from langchain_classic.chains import ConversationChain
from langchain_classic.memory import (
    ConversationBufferMemory,
    ConversationSummaryMemory,
    ConversationBufferWindowMemory,
    ConversationTokenBufferMemory,
)

import tiktoken
from dotenv import load_dotenv
load_dotenv()
```

```python
llm = ChatOpenAI(temperature=0.0)
```

ConversationBufferMemory

```python
con_chain = ConversationChain(
    llm=llm,
    verbose=True,
    memory=ConversationBufferMemory()
)
con_chain.prompt.template
```

```python
con_chain("Good Morning!")
```

```python
con_chain("My Name is Ajay, I  am 28 years old")
```

```python
con_chain("How to design ML pipeline for house prediction")
```

```python
con_chain.predict(input="I stay in Pune MH India")
```

```python
con_chain.memory.buffer
```

```python
con_chain.predict(input="What is my name and my  location")
```

ConversationBufferWinfowMemory

```python
conv = ConversationChain(
    llm=llm,
    verbose=True,
    memory = ConversationBufferWindowMemory(k=3)
)
conv.prompt.template
```

```python
conv.predict(input="Hi, Good Morning! My name is Kumar")
```

```python
conv.predict(input="what is 2+12")
```

```python
conv.predict(input="I stay in Mumbai near sea facing balcony")
```

```python
conv.memory.buffer
```

```python
conv.predict(input="What is my Name")
```

ConversationSummaryMemory

```python
conv = ConversationChain(
    llm = llm,
    verbose=True,
    memory=ConversationSummaryMemory(llm=llm)
)
conv.prompt.template
```

```python
conv.memory.prompt.template
```

```python
conv.predict(input="Hi. Good Morning, my name is Tola")
```

```python
conv.predict(input="I stay in Bombay")
```

```python
conv.memory.buffer
```

```python
conv.predict(input="what is my name")
```

ConversationTokenBufferMemory

```python
conversation = ConversationChain(
    llm=llm,
    verbose=True,
    memory=ConversationTokenBufferMemory(llm=llm, max_token_limit=60),
)

conversation.prompt.template
```

```python
conversation.predict(input="Good morning AI!")
```

```python
conversation.predict(input="My Name is sharath")
```

```python
conversation.predict(input="I stay in hyderabad")
```

```python
conversation.memory.buffer
```

```python
conversation.predict(input="What is my name?")
```
