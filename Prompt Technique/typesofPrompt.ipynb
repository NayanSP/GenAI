Prompt Chaining
- involves linking multiple prompts together, where the output of one prompt serves as the input for the next. 
- This technique is ideal for multistep processes.

```python
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

load_dotenv()

llm = ChatOpenAI(
    model="gpt-4.1-mini",
    temperature=0
)

parser = StrOutputParser()

# Prompt 1: Summarization
summary_prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a document summarizer."),
    ("human", "Summarize the following text:\n\n{text}")
])

summary_chain = summary_prompt | llm | parser

# Prompt 2: Extract Key Points
keypoint_prompt = ChatPromptTemplate.from_messages([
    ("system", "You are an expert analyst."),
    ("human", "Extract the top 5 key points from:\n\n{summary}")
])

keypoint_chain = keypoint_prompt | llm | parser

# Prompt 3: Create Final Report
report_prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a report writer."),
    ("human", """
Create a professional report.

Summary:
{summary}

Key Points:
{points}
""")
])

report_chain = report_prompt | llm | parser

text = """
Artificial Intelligence is transforming healthcare by improving diagnostics,
drug discovery, and personalized medicine. However, ethical concerns such as
privacy, bias, and transparency remain major challenges.
"""

summary = summary_chain.invoke({"text": text})

points = keypoint_chain.invoke({
    "summary": summary
})

final_report = report_chain.invoke({
    "summary": summary,
    "points": points
})

print(final_report)
```
