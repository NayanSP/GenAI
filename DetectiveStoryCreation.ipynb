```python
import asyncio
import edge_tts

async def voice(TEXT):
    communicate = edge_tts.Communicate(
        TEXT,
        "en-US-AriaNeural"
    )
    await communicate.save("output1.mp3")
```

```python
prompt = PromptTemplate(
    template="""
You are a writer who can write any story based on the topic and generes.

You will be provided with the idea, th number of line it should contain, and most important the genere it should follow.

You have to write on {topic} for exact {number} lines to be read for {types} genere.
below are the requirment:
1. the blog can be read and heared by people above age group of 10
2. the blog should not hurt anyone nationalist and culture sentiment in the world.
3. the blog should be complete fictional
4. the person in the blog should be around 5 and not more than that
5. the blog should contain excitment, should give adreline rush to the audiance 
6. the blog should leave lasting remarks on the audiance 
""",
input_variables=['topic', 'number','types']
)
```

```python
llm = ChatOpenAI(
   
    temperature=1.7,
    top_p=0.8
)
```

```python
response = llm.invoke(
    prompt.format(
        topic='getting file in from a blockchain, alraming the owner of the theft, mocking the owner for keeping password in front of his eyes.',
        number='100',
        types = 'mystery, detective, crime, romance'
    )
    )
```
