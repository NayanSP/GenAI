```python
import streamlit as st
from langchain_core.prompts import PromptTemplate
from langchain_community.llms import CTransformers
from langchain_openai import ChatOpenAI
from dotenv import load_dotenv
load_dotenv()
```

```python
import asyncio
import edge_tts

async def voice(TEXT):
    communicate = edge_tts.Communicate(
        TEXT,
        "en-US-AriaNeural"
    )
    await communicate.save("output1.mp3")
```

```python
prompt = PromptTemplate(
template="""
You are an award-winning fiction writer and storyteller.

Your task is to write a captivating {types} story based on the topic: "{topic}".

Requirements:

1. The story must be completely fictional.
2. The story should be suitable for audiences aged 10 and above.
3. The story must respect all cultures, nationalities, and communities.
4. The story should contain excitement, suspense, emotions, and memorable moments.
5. The story should maintain the style and tone of a {types} genre throughout.
6. The story should be immersive and suitable for text-to-speech narration.
7. The story should contain approximately {word_count} words, resulting in an audio narration length of about 5–7 minutes.
8. Use simple, natural, conversational language that sounds pleasant when spoken aloud.
9. Avoid bullet points, lists, headings, or chapter breaks.
10. End with a powerful and memorable conclusion.
11. Output only the story text.
12. Use short and medium-length sentences suitable for text-to-speech narration.
13. Include descriptive scenes and dialogue.
14. Maintain pacing throughout the story.
15. Target 800–1000 words to achieve 5–7 minutes of narration.
16. Avoid abrupt scene changes.
17. Write in a natural storytelling style as if a narrator is telling the story aloud.

Topic: {topic}
Genre: {types}
Target Word Count: {word_count}
""",
input_variables=["topic", "types", "word_count"]
)
```

```python
llm = ChatOpenAI(
    model="gpt-5.5",
    temperature=1.2
)
```

```python
response = llm.invoke(
    prompt.format(
        topic="""
A murder in railway station at mindnight, with only clue of toothpick and shoes of the dead person.
His eyes were looking into the ad present in the billboard of railway station entrance.
the decetive marlin with her sidekick andy starts the investigation.
they finds clue with a revenge story of the murderer and dead person
""",
        word_count="1500",
        types="Mystery Detective Crime Thriller"
    )
)
```

```python
with open('story1.txt', "w", encoding="utf-8") as f:
    f.write(response.content)

print(f"Saved to story.txt")
```
